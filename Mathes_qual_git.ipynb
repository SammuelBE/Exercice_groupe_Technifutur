```python
import numpy as np
import pandas as pd
import urllib.parse

import requests as rq
from io import StringIO
from sqlalchemy import create_engine
```

```python
# Dictionnaire vide pour stocker les DataFrames
data_tennis = {}

# Boucle de 1999 à 2017 inclus
for annee in range(1999, 2018):
    # Construction dynamique du chemin du fichier
    chemin = rf"C:\Users\ST1803-11\Downloads\Tennis Data\atp_matches_qual_chall_{annee}.csv"
    
    # Lecture et stockage dans le dictionnaire
    data_tennis[annee] = pd.read_csv(chemin)

# Exemple pour afficher les données de l'année 2005 :
# print(data_tennis[2005])
```

```python
for annee in range(1999, 2010):
    data_tennis[annee] = data_tennis[annee].drop(columns=['minutes', 'w_ace', 'w_df', 'w_svpt', 'w_1stIn', 'w_1stWon', 'w_2ndWon', 'w_SvGms', 'w_bpSaved', 'w_bpFaced', 'l_ace', 'l_df', 'l_svpt', 'l_1stIn', 'l_1stWon', 'l_2ndWon', 'l_SvGms', 'l_bpSaved', 'l_bpFaced'])
```

```python
data_tennis[1999]

# 2. Configurer la connexion à SQL Server (SSMS)
# Adaptez le 'Server' et le 'Database' selon votre environnement local
server_name = r'TF1803-VM11\DATAVIZ' 
database_name = 'Maches_qual'

params = urllib.parse.quote_plus(
    f"DRIVER={{ODBC Driver 17 for SQL Server}};"
    f"SERVER={server_name};"
    f"DATABASE={database_name};"
    f"Trusted_Connection=yes;"
)

engine = create_engine(f"mssql+pyodbc:///?odbc_connect={params}")

# 3. Exporter le DataFrame vers SQL Server
for annee in range(1999, 2018):
    # Correction : On génère le nom de la table sous forme de texte (string)
    nom_table_sql = f'Matches_qual{annee}'

    df_annee = data_tennis[annee]
    
    # Note : Assurez-vous que la variable 'df' correspond bien aux données de cette année spécifique ici
    df_annee.to_sql(
        name=nom_table_sql, 
        con=engine, 
        if_exists='replace', 
        index=False          
    )

    print(f"Les données ont bien été importées dans la table {nom_table_sql} de SSMS !")
```
